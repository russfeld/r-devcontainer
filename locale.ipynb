```r
library(tidyverse)

raw_data <- read.csv("ris_locale.csv", header = TRUE)
locale <- raw_data$locale
ris <- raw_data$ris

shapiro.test(raw_data$ris)

pairwise.t.test(raw_data$ris, raw_data$locale, pool.sd = TRUE, p.adjust.method = "bonferroni")
pairwise.wilcox.test(raw_data$ris, raw_data$locale, pool.sd = TRUE, p.adjust.method = "bonferroni")
```
